# BBO Capstone Project Notebook

This notebook presents the recorded optimisation history for the Imperial College Black-Box Optimisation (BBO) Capstone Project. The goal was to maximise eight hidden objective functions using a limited number of query opportunities.

The notebook is intended as a clear presentation of the method and results rather than a production-grade optimiser.

## Approach

The optimisation strategy evolved over multiple rounds:

1. broad exploration of the search space;
2. local refinement around promising points;
3. pattern-based search using previous input-output pairs;
4. balancing exploration and exploitation per function;
5. documenting score trends and final lessons learned.
